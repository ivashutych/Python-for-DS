1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

```python
def func(x):
    return 10*x - 5
```

```python
def df(func, x, h = 1e-8):
    return (func(x+h) - func(x)) / h
```

```python
df(func, 3)
```

```python
x = np.linspace(0, 10, 10)
f = func(x)
df_values = df(func,x)
# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, f, label='Функція', color='blue')
plt.plot(x, df_values, label="Похідна", color='green')

plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

З кожною годиною оцінка покращується на 10 балів, не змінюючи "швидкість".

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')

f_sym = sp.sin(x)
df_sym = sp.diff(f_sym, x)
print("Функція:", f_sym)
print("Похідна:", df_sym)
```

```python
#Функція побудови графіку
def build_plot(f_sym, df_sym, a, b, c, xlabel = 'x', ylabel='y'):
    x = sp.symbols('x')
    f_func = sp.lambdify(x, f_sym, 'numpy')
    df_func = sp.lambdify(x, df_sym, 'numpy')

    x_values = np.linspace(a, b, c)
    f_values = f_func(x_values)
    df_values = df_func(x_values)

    plt.figure(figsize=(10, 5))
    plt.plot(x_values, f_values, label='Функція', color='blue')
    plt.plot(x_values, df_values, label="Похідна", color='green')
    
    plt.title("Графік функції та її похідної")
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend()
    plt.grid(True)
    
    plt.show()
```

```python
build_plot(f_sym, df_sym, 0, 10, 1000)
```

```python
x = sp.symbols('x')

f_sym_2 = sp.ln(x)
df_sym_2 = sp.diff(f_sym_2, x)
print("Функція:", f_sym_2)
print("Похідна:", df_sym_2)
```

```python
build_plot(f_sym_2, df_sym_2, 0.5, 100, 1000)
```

```python
x = sp.symbols('x')

f_sym_3 = x**2-5*x+1
df_sym_3 = sp.diff(f_sym_3, x)
print("Функція:", f_sym_3)
print("Похідна:", df_sym_3)
```
